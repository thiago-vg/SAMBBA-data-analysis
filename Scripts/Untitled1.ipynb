```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
BASE_URL = "https://data.ceda.ac.uk/badc/faam/data/2012/b747-oct-01/core_processed"

def list_files(url):
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Failed to access {url}")
        return []
    
    soup = BeautifulSoup(response.text, "html.parser")
    file_urls = []
    
    for link in soup.find_all("a"):
        href = link.get("href")
        if href and not href.startswith("?") and not href.startswith("/"):
            file_urls.append(f"{url}/{href}")
    
    return file_urls
```

```python
files = list_files(BASE_URL)
for file in files:
    print(file)
```

```python
# Define base URL and output directory
BASE_URL = "https://dap.ceda.ac.uk/badc/faam/data/2012/"
OUTPUT_DIR = "/home/jovyan/SAMBBA_data/"  # Change this to your desired directory
TOKEN = "your_authentication_token_here"  # Replace with your actual token

# List of subdirectories to download
url_suffixes = ["b729-aug-28"]  # Add more as needed

# Ensure the output directory exists
os.makedirs(OUTPUT_DIR, exist_ok=True)

def download_data():
    for suffix in url_suffixes:
        url = BASE_URL + suffix + "/"

        # Get the directory listing
        headers = {"Authorization": f"Bearer {TOKEN}"}
        response = requests.get(url, headers=headers)

        if response.status_code != 200:
            print(f"Failed to access {url} (status code: {response.status_code})")
            continue

        # Extract filenames from the directory listing
        file_list = extract_file_links(response.text, url)

        for file_url in file_list:
            file_name = os.path.join(OUTPUT_DIR, os.path.basename(file_url))
            print(f"Downloading: {file_url}")

            file_response = requests.get(file_url, headers=headers, stream=True)
            
            # Save the file
            with open(file_name, "wb") as file:
                for chunk in file_response.iter_content(chunk_size=8192):
                    file.write(chunk)
            
            print(f"Saved: {file_name}")

def extract_file_links(html_text, base_url):
    """Extracts file URLs from an HTML directory listing"""
    soup = BeautifulSoup(html_text, "html.parser")
    links = []

    for link in soup.find_all("a"):
        href = link.get("href")
        if href and not href.endswith("/"):  # Avoid directories
            links.append(base_url + href)

    return links

# Run the download function
download_data()
```
