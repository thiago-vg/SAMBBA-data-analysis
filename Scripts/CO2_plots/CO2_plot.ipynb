```python
def geometric_mean_diameter(D_lower, D_upper, concentrations):
    """
    Compute geometric mean diameter (µm) from channel boundaries and concentrations.
    """
    # Bin midpoints in log-space
    D_mid = np.sqrt(D_lower * D_upper)

    # Ensure concentrations is numpy array and mask invalids
    concentrations = np.array(concentrations, dtype=float)
    mask = (concentrations > 0) & (concentrations != 9999)

    if not np.any(mask):
        #print('retornou')
        return np.nan  # no valid data

    conc = concentrations[mask]
    D_mid = D_mid[mask]

    # Weighted geometric mean
    log_Dg = np.nansum(conc * np.log(D_mid)) / np.nansum(conc)
    return np.exp(log_Dg)
```

```python
def convert_time_to_datetime_sp2(dsf, start_date_str):
    """
    Convert the 'Time_start_UTC' variable (seconds from 00:00 UTC) into datetime format.
    
    Parameters:
    - ds (xarray.Dataset): The dataset containing 'Time_start_UTC'
    - start_date_str (str): Start date in "YYYY MM DD" format
    
    Returns:
    - xarray.Dataset: Dataset with a new 'datetime' coordinate
    """
    # Convert start_date string to datetime object
    start_date = datetime.strptime(start_date_str, "%Y%m%d")
    
    # Extract 'Time_start_UTC' variable (seconds from 00:00 UTC)
    time_seconds = dsf["Time_UTC"].values
    
    # Convert seconds to actual datetime
    datetime_values = np.array([start_date + timedelta(seconds=int(sec)) for sec in time_seconds])

    # Assign datetime as a coordinate in the dataset
    dsf = dsf.assign_coords(datetime=("time", datetime_values))

    return dsf
    
def convert_time_to_datetime_ams(dsf, start_date_str):
    """
    Convert the 'Time_start_UTC' variable (seconds from 00:00 UTC) into datetime format.
    
    Parameters:
    - ds (xarray.Dataset): The dataset containing 'Time_start_UTC'
    - start_date_str (str): Start date in "YYYY MM DD" format
    
    Returns:
    - xarray.Dataset: Dataset with a new 'datetime' coordinate
    """
    # Convert start_date string to datetime object
    start_date = datetime.strptime(start_date_str, "%Y%m%d")
    
    # Extract 'Time_start_UTC' variable (seconds from 00:00 UTC)
    time_seconds = dsf["Time_mid_UTC"].values
    
    # Convert seconds to actual datetime
    datetime_values = np.array([start_date + timedelta(seconds=int(sec)) for sec in time_seconds])

    # Assign datetime as a coordinate in the dataset
    dsf = dsf.assign_coords(datetime=("time", datetime_values))

    return dsf
    
def convert_time_to_datetime_neph(dsf, start_date_str):
    
    # Convert start_date string to datetime object
    start_date = datetime.strptime(start_date_str, "%Y%m%d")

    key_options = ["neph_spm", "NEPH_SPM"]  # Possible variations of the key
    matching_key = next((key for key in key_options if key in dsf), None)
    
    if matching_key:
        time_seconds_t = dsf[matching_key].values
    else:
        raise KeyError("Variable 'neph_spm' not found in dataset.")
    
    # Extract 'Time_start_UTC' variable (seconds from 00:00 UTC)
    #time_seconds = dsf["neph_spm"].values
    
    # Convert seconds to actual datetime
    datetime_values = np.array([start_date + timedelta(seconds=int(sec)) for sec in time_seconds_t])

    dsf = dsf.assign_coords(time=dsf['neph_spm'])
    # Assign datetime as a coordinate in the dataset
    dsf = dsf.assign_coords(datetime=("time", datetime_values))
    #dsf["Times"] = (("time",), datetime_values)
    #dsf["datetime"] = (("time",), datetime_values)
    #dsf = dsf.set_coords('neph_spm')

    return dsf
def convert_time_to_datetime_co2(dsf, start_date_str):
    """
    Convert the 'Time_sec' variable (seconds from 00:00 UTC) into datetime format.
    
    Parameters:
    - ds (xarray.Dataset): The dataset containing 'Time_sec'
    - start_date_str (str): Start date in "YYYY MM DD" format
    
    Returns:
    - xarray.Dataset: Dataset with a new 'datetime' coordinate
    """
    # Convert start_date string to datetime object
    start_date = datetime.strptime(start_date_str, "%Y%m%d")
    
    # Extract 'Time_start_UTC' variable (seconds from 00:00 UTC)
    time_seconds = dsf["Time_sec"].values
    
    # Convert seconds to actual datetime
    datetime_values = np.array([start_date + timedelta(seconds=int(sec)) for sec in time_seconds])

    # Assign datetime as a coordinate in the dataset
    dsf = dsf.assign_coords(datetime=("time", datetime_values))

    return dsf
```

```python
def file_ams_sp2_treatment_v3(aux_ds_sp2, aux_ds_ams):
    """
    Process AMS and SP2 datasets:
    - Filter invalid values
    - Exclude points where uncertainty > 0.5 * value
    - Resample to 1 min averages
    - Propagate uncertainties
    - Compute ammonium sulfate and nitrate mass concentrations
    """
    variables = ["SO4", "NH4", "ORG", "NO3", "Chl"]
    variables_unc = [f"{var}_uncertainty" for var in variables]

    resampled_vars = {}
    resampled_uncs = {}

    for var, var_unc in zip(variables, variables_unc):
        # Mask invalid data
        valid_mask = (aux_ds_ams[var] != 9999999) & (aux_ds_ams[var_unc] != 9999999)

        # Additional restriction: only accept if uncertainty < 0.5 × value
        valid_mask &= aux_ds_ams[var_unc] < 0.5 * aux_ds_ams[var]

        data = aux_ds_ams[var].where(valid_mask)
        unc = aux_ds_ams[var_unc].where(valid_mask)

        # Resample (1 min mean)
        resampled_vars[var] = data.resample(datetime="1min").mean()

        # Count of valid points
        counts = data.resample(datetime="1min").count()

        # Propagate uncertainty: σ_mean = sqrt(sum(σ_i²)) / N
        sum_unc_sq = (unc ** 2).resample(datetime="1min").sum()
        resampled_uncs[var_unc] = np.sqrt(sum_unc_sq) / counts

    # Combine all processed variables
    resampled_ds_ams = xr.Dataset({**resampled_vars, **resampled_uncs})

    # --- Add derived variables ---
    # (NH4)2SO4 = 1.375 * SO4
    resampled_ds_ams["NH4_2SO4"] = 1.375 * resampled_ds_ams["SO4"]

    # NH4NO3 = 1.292 * NO3
    resampled_ds_ams["NH4NO3"] = 1.292 * resampled_ds_ams["NO3"]

    # Process BC_mass
    valid_BC_mass_mask = aux_ds_sp2["BC_mass"] != 9999999
    filtered_ds_BC_mass = aux_ds_sp2["BC_mass"].where(valid_BC_mass_mask) * 1e-3  # Convert to µg/m³
    resampled_ds_sp2 = filtered_ds_BC_mass.resample(datetime="1min").mean()

    return resampled_ds_ams, resampled_ds_sp2
```

```python
def identify_constant_altitude(
    df,
    alt_col='ALT',
    window='5min',
    std_threshold=30  # meters
):
    """
    Identify constant-altitude flight segments.

    Returns a boolean Series.
    """
    alt_std = (
        df[alt_col]
        .rolling(window=window, center=True)
        .std()
    )

    return alt_std < std_threshold
```

```python
def estimate_background_multi(
    df,
    tracers,
    mask_constant_altitude,
    percentile=5
):
    """
    Estimate background values for multiple tracers.
    """
    bg = {}

    for tracer in tracers:
        data = df.loc[mask_constant_altitude, tracer].dropna()
        bg[tracer] = np.percentile(data, percentile) if not data.empty else np.nan

    return bg
```

```python
def classify_plume_multi(
    df,
    bg,
    enhancement_factors,
    enhancement_offsets=None,
    min_tracers=2
):
    """
    Classify in-plume using multiple tracers.

    min_tracers: minimum number of enhanced tracers required
    """
    enhanced = []

    for tracer, factor in enhancement_factors.items():
        bg_val = bg.get(tracer, np.nan)

        if np.isnan(bg_val):
            enhanced.append(pd.Series(False, index=df.index))
            continue

        if enhancement_offsets and tracer in enhancement_offsets:
            threshold = bg_val + enhancement_offsets[tracer]
        else:
            threshold = bg_val * factor

        enhanced.append(df[tracer] > threshold)

    enhanced_count = pd.concat(enhanced, axis=1).sum(axis=1)

    return enhanced_count >= min_tracers
```

```python
def identify_background_and_plume_multi(
    df,
    flight_col='Flight',
    tracers=('CO', 'BC', 'ORG'),
    alt_col='ALT',
    alt_window= '5min',
    alt_std_threshold=30,
    bg_percentile=5,
    enhancement_factors=None,
    enhancement_offsets=None,
    min_tracers=2
):
    """
    Identify background and in-plume measurements using multiple tracers.
    """

    if enhancement_factors is None:
        enhancement_factors = {
            'CO': 1.2,
            'BC': 1.5,
            'ORG': 1.5
        }

    df = df.copy()

    df['constant_altitude'] = False
    df['in_plume'] = False
    df['is_background'] = False

    # background values stored per tracer
    for tracer in tracers:
        df[f'{tracer}_bg'] = np.nan

    for flight, dff in df.groupby(flight_col):

        # Constant altitude mask
        const_alt = identify_constant_altitude(
            dff,
            alt_col=alt_col,
            window=alt_window,
            std_threshold=alt_std_threshold
        )

        # Background estimation
        bg = estimate_background_multi(
            dff,
            tracers=tracers,
            mask_constant_altitude=const_alt,
            percentile=bg_percentile
        )

        # Plume classification
        plume_flag = classify_plume_multi(
            dff,
            bg=bg,
            enhancement_factors=enhancement_factors,
            enhancement_offsets=enhancement_offsets,
            min_tracers=min_tracers
        )

        # Store results
        df.loc[dff.index, 'constant_altitude'] = const_alt
        df.loc[dff.index, 'in_plume'] = plume_flag
        df.loc[dff.index, 'is_background'] = (~plume_flag) & const_alt

        for tracer in tracers:
            df.loc[dff.index, f'{tracer}_bg'] = bg[tracer]

    return df
```

```python
def plot_tracer_timeseries_flight(
    df,
    flight_id,
    tracer='ORG',
    flight_col='Flight',
    figsize=(12, 4)
):
    """
    Plot time series of one tracer for a given flight,
    colored by plume/background flags.
    """

    dff = df[df[flight_col] == flight_id]

    if dff.empty:
        raise ValueError(f"No data found for flight {flight_id}")

    fig, ax = plt.subplots(figsize=figsize)

    # Background
    bg = dff[dff['is_background']]
    ax.scatter(
        bg.index,
        bg[tracer],
        s=12,
        color='tab:blue',
        label='Background',
        alpha=0.7
    )

    # In plume
    plume = dff[dff['in_plume']]
    ax.scatter(
        plume.index,
        plume[tracer],
        s=12,
        color='tab:red',
        label='In plume',
        alpha=0.7
    )

    # Other / transition
    other = dff[~(dff['in_plume'] | dff['is_background'])]
    ax.scatter(
        other.index,
        other[tracer],
        s=10,
        color='tab:gray',
        label='Other',
        alpha=0.4
    )

    ax.set_ylabel(tracer)
    ax.set_xlabel('Time')
    ax.set_title(f'Flight {flight_id} – {tracer} time series')
    ax.legend()
    ax.grid(True)

    plt.tight_layout()
    plt.show()
```
