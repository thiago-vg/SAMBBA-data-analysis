```python
# Define the URL and output directory
URL = "https://dap.ceda.ac.uk/badc/faam/data/2012/b731-sep-14/"
OUTPUT_DIR = '/home/jovyan/SAMBBA_data/'  # Change this to your desired directory
```

```python
# Base URL and output directory
BASE_URL = "https://dap.ceda.ac.uk/badc/faam/data/2012/"
```

```python
# List of URL suffixes
url_suffixes = [
    "b729-aug-28"#, "b730-sep-04", "b731-sep-14", "b732-sep-15", "b733-sep-16", 
#    "b734-sep-18", "b735-sep-19", "b736-sep-19", "b737-sep-20", "b738-sep-22", 
#    "b739-sep-23", "b740-sep-25", "b741-sep-26", "b742-sep-27", "b743-sep-27", 
#    "b744-sep-28", "b745-sep-28", "b746-sep-29", "b747-oct-01", "b748-oct-02", 
#    "b749-oct-03", "b750-oct-03"
]
```

```python
# Loop through each suffix and download the data
def download_data():
    for suffix in url_suffixes:
        url = BASE_URL + suffix + "/"
        command = [
            "wget",
            "-e", "robots=off",
            "--mirror",  # Enable full mirroring
            "--no-parent",  # Prevent going up directories
            "-r",
            "--accept", "nc,csv",  # Only download .nc and .csv files
            "--header", f"Authorization: Bearer {TOKEN}",  # Add authentication header
            "-P", OUTPUT_DIR,  # Specify output directory
            url  # Target URL
        ]
        
        try:
            subprocess.run(command, check=True)
            print(f"Download completed successfully for {suffix} and saved to {OUTPUT_DIR}")
        except subprocess.CalledProcessError as e:
            print(f"Error during download of {suffix}: {e}")
```

```python
# Execute the download function
download_data()
```

```python
# Function to download data using requests
def download_data():
    for suffix in url_suffixes:
        url = BASE_URL + suffix + "/"
        response = requests.get(url, headers=headers)
        
        if response.status_code == 200:
            file_path = os.path.join(OUTPUT_DIR, f"{suffix}.html")  # Default saving as HTML for directory listing
            with open(file_path, "wb") as file:
                file.write(response.content)
            print(f"Downloaded: {suffix} -> {file_path}")
        else:
            print(f"Failed to download {suffix}. HTTP Status Code: {response.status_code}")
```

```python
# Execute the download function
download_data()
```

```python
# Function to download data using wget with authentication
def download_data():
    for suffix in url_suffixes:
        url = BASE_URL + suffix + "/"
        command = f'wget -e robots=off --mirror --no-parent -r "{url}" --header "Authorization: Bearer {TOKEN}" -P "{OUTPUT_DIR}"'
        os.system(command)
        print(f"Downloading: {url}")

# Execute the download function
download_data()
```
